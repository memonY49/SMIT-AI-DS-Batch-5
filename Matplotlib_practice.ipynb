```python
plt.plot(x, y, label = 'Blue line', c = '#04cf9c', marker = '^',mfc = 'r',mec = 'r')
plt.title("X vs Y", pad=40)
plt.text(0,15,"data")
plt.xlabel("X-Axis",labelpad=30)
plt.ylabel("Y-Axis")
plt.legend(loc = (1.1,0.9))
plt.show()
```

```python
y1 =[]
y2 =[]
x = range(-100,101,10)
for i in x: y1.append(i**2) 
for i in x: y2.append(-i**2) 
```

```python
plt.plot(x,y1,label = 'Y1', ls = '--')
plt.plot(x,y2,label = 'Y2')
plt.axhline(y=0, c = 'r', alpha = 0.5)
plt.axvline(x=0, c = 'r', alpha = 0.5)
plt.xlim(-200,200)
plt.ylim(-20000,20000)


plt.legend(loc = (1,1))
```

```python
values = [5, 6, 3, 7, 2]
names  = ["A", "B", "C", "D", "E"]
barchart = plt.bar(names,values,label = names,color = ['r','b','y','brown','orange'],alpha=0.5)
plt.legend()
plt.bar_label(barchart)
# plt.figure(figsize=(10,10))

plt.savefig("mybarchart.png")
plt.show()
```

```python
dataset = pd.read_csv("updated_student_data.csv")
```

```python
dataset
```

```python
grade_count = dataset.grades.value_counts()
```

```python
plt.bar(grade_count.index, grade_count.values)
```

```python
grade_count.plot.bar()
```

```python
grouped_data = dataset.groupby('class')
```

```python
for cl in dataset['class'].unique():
    cl_data = grouped_data.get_group(cl)
    plt.title(cl)
    cl_data.grades.value_counts().plot.bar()
    plt.show()
```

```python
grouped_data['grades'].value_counts().plot.bar()
plt.legend()
```

```python
class_grade_count = grouped_data['grades'].value_counts()
```

```python
len(class_grade_count)
```

```python
str_class_grade_count = [str(label) for label in class_grade_count.index]
```

```python
plt.figure(figsize=(12,12))
plt.bar(str_class_grade_count,class_grade_count.values, color = [ f"#fa0{505+10*n}" for n in range(1,len(class_grade_count)+1)],width = 1)
plt.xticks(rotation = 90)
```
