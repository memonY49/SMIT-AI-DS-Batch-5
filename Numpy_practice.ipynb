```python
marks = np.array([[40,50,80,90],
                 [40,55,89,70],
                 [30,60,80,60],
                 [40,50,80,90],
                 [40,55,89,70],
                 [30,60,80,60]])
total_marks = marks.sum(axis=1)
per = (total_marks/400)*100
ranking = per.argsort()[::-1]

for rank, index in enumerate(ranking):
    print("student no:",index+1)
    print("rank no:",rank+1)
    print("total marks:",total_marks[index])
    print("percentage:",per[index])
    print("*"*20)

```

# Task1
- Find Max marked student per subject
- Find Total Obtain Marks for every student
- total percentage for every student (500/obtain)*100
- Find Grade for every student
- Find the Rank for every student

```python
student_data = genfromtxt('student.csv',delimiter=',',dtype=str)
```

```python
subject = student_data[0]
```

```python
student_data = np.delete(student_data,0, axis=0).astype(int)
```

```python
subject
```

```python
max_std_subject = student_data.argmax(axis=0)
```

```python
total_obtain = student_data.sum(axis=1)
total = len(subject)*100
```

```python
per = (total_obtain/total)*100
```

```python
rank = per.argsort()[::-1]
```

```python
grades = np.where(per>80,'A',
             np.where(per>70,'B',
                 np.where(per>60,'C',
                         np.where(per>50,'D','F'))))
```

```python
max_std_subject
```

```python
per
```

```python
rank
```

```python
for r, v in enumerate(rank):
    print(f"Position: {r+1} \nstudent no: {v+1}")
    to_subjects = max_std_subject[max_std_subject == v]
    print(f"Top Subjects: {len(to_subjects)}")
    print(f"Obtain Marks: {total_obtain[v]}")
    print(f"Per: {per[v]}")
    print(f"Grade: {grades[v]}")
    print("*"*20)
```
