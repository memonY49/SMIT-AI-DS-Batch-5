```python
students = pd.read_csv("students_five_classes.csv")
```

```python
students["class"] = students.roll_no.str.split('/').str[1]
```

```python
students
```

```python
students.info()
```

```python
students['obtain'] = students.iloc[:,2:-1].sum(axis = 1)
```

```python
students['total'] = 100
```

```python
students['per'] = (students.obtain/students.total)*100
```

```python
students
```

```python
def grade(x):
    if x>=80:
        return 'A'
    elif x>=60:
        return 'B'
    elif x>=40:
        return 'C'
    elif x<40:
        return 'F'
        
```

```python
students['grades'] = students.per.apply(grade)
```

```python
students.to_csv("updated_student_data.csv")
```
